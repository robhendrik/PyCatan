## Overview

This index notebook summarizes the tutorial notebooks included in the `docs/` folder and links their main ideas, code snippets, and visual assets. Follow the sections below in order — each tutorial is runnable from the repository root and uses the `src/` package for the PyCatan implementation.

### What this repo demonstrates

- Introduction: what we want to achieve and how we simplify Catan for research/education purposes.
- Tools: Python, Keras/TensorFlow, AWS (S3 for artifacts), and optional OpenAI integrations for verbalization/personas.
- Player personalities: players are given short 
personas
 (e.g. aggressive, conservative) — some are inspired by fictional AIs to illustrate voice and strategy variation.
- Models: each player uses three separate Keras models (Setup, Gameplay, Trade). We typically bootstrap them from imitation-learning datasets and refine with PPO-style reinforcement learning.
- Evaluation: how we measure performance (tournaments, KL/JS divergence diagnostics, flip rates, win% against baselines).
- Comparison: random, heuristic, imitation, and RL players are compared in evaluation tournaments; results are summarized as tables and charts in the tutorials.
- Video/visuals: the `http/` folder contains images used in tutorial pages and a short demo video reference.
